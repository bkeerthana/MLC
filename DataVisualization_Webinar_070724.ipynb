**Dot Plot**

    Representation

     - Graphical representation of data using dots.

    Also called as

    - strip plot
    - dot chart

    Types
      
    - Wikinson Dot Plot
    
    - Cleveland Dot Plot

    Cleveland Dot Plot

    - created by William S.Cleveland.
    - Use position to plot the value.
    - similar to bar graph.



    Wikinson Dot Plot

    - Created by Leland Wikinson.
    - prevent the dots on the plot from overlapping
    - Displays a distribution of continuous data points.
    - Similar to histogram, but it display individual data points instead of bins.

    Application

    - To display the distribution of data
    - To cluster the value
    - To find outliers

    Understanding Data

    -  visualize the shape and spread of data
    -  comparing frequency distributions - indicates how often values in a dataset occurs.

    Clustering

    - group the data into different cluster

    Outliers

    - Identify any abnormal data
    
    Interpreting the Dot Plot
    - What is the minimum duration
    - What is the maximum duration

    used when
    - relatively small data sets where values fall into a number of discrete categories.

    Drawback
    - Need to count the total number of data elements in each stack of dots.
    - Difficult to construct and interpret data set for many points.

Wilkinson Dot Plot

The dots are plotted against their actual data values that are on the horizontal scale.
If there are identical data values, the dots are “piled” on top of each other.

Thus, to draw a dot plot, count the number of data points falling in each data value and draw a stack of dots that corresponds to the number of items in each data value.

```python
data  = [4,7,3,6,1,8,9,2,4,7,4,3]
pos = []
keys = {} # this dict will help to keep track ...

# this loop will give us a list of frequencies to each number
for num in data:
   if num not in keys:
      keys[num] = 1
      pos.append(1)
   else:
      keys[num] += 1
      pos.append(keys[num])


print(pos)

plt.plot(data, pos,'o')
plt.xlabel('No. of Hours')
plt.ylabel('No. of Students')
plt.show()
```

**Subplots**

Often times we might want to plot multiple plots within the same figure. For example, we might want to perform a side by side comparison of the box plot with the line plot of China and India's immigration.

To visualize multiple plots together, we can create a figure (overall canvas) and divide it into subplots, each containing a plot. With subplots, we usually work with the artist layer instead of the scripting layer.

Typical syntax is :

fig = plt.figure() # create figure
    ax = fig.add_subplot(nrows, ncols, plot_number) # create subplots
Where

nrows and ncols are used to notionally split the figure into (nrows * ncols) sub-axes,
plot_number is used to identify the particular subplot that this function is to create within the notional grid. plot_number starts at 1, increments across rows first and has a maximum of nrows * ncols as shown below.
