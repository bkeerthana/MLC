Confusion Matrix

image.png





Two components
   - correctness
   - class label (predict)

TP → Correctly predicted positive

TN → Correctly predicted negative

FP → Incorrectly predicted positive

FN → Incorrectly predicted negative

Error
   - Type I  (False Alarm)
   - Type II (Miss)

image.png

Accuracy → Overall proportion of correct predictions (both positive and negative)

Error Rate → Overall proportion of incorrect predictions

Precision → Out of predicted positives, how many are actually correct

Recall → Out of actual positives, how many were correctly identified

F1-score → Balanced measure combining precision and recall (harmonic mean)

Accuracy

→ Use when classes are balanced and all errors are equally important
Example: General classification where positive/negative are equally likely

Error Rate

→ Use when you want to explicitly measure overall mistakes
Example: System reliability evaluation (how often the model fails)

Precision

→ Use when false positives are costly
Example: Spam filter → don’t mark legitimate emails as spam
 Focus: “When I say attack, I must be correct”

Recall

→ Use when false negatives are costly
Example: DDoS attack detection → don’t miss actual attacks

 Focus: “I must catch all attacks”

 F1-score

→ Use when you need a balance between precision and recall (especially in imbalanced data)
Example: Cybersecurity datasets where attacks are rare but important

If missing attack is dangerous → Recall

If false alarm is costly → Precision

If both matter → F1-score

If data is balanced → Accuracy
